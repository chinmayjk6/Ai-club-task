```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('atlantis_citizens_final.csv')
df['commute_index'] = df['District_Name'] != df['Work_District']
# Group by 'District Name' and count how many rows have commute_index == True
commuting_counts = df[df['commute_index'] == True].groupby('District_Name').size().reset_index(name='Commuter Count')
commuting_counts
```
